# Figure 1 — The construction

Artwork for Figure 1, drawn with matplotlib in millimetre coordinates (`network/artwork_fig1.py`).

- **A.** Microstates (gray dots) and their transitions (faint arrows) form a jump process. Coarse observables are overlapping groups of microstates: the exposed coordinate $\mu$, the boundary $s$ and the slow structure $W$. Funnels carry them to the bundle: the fast coordinates to a point on a fiber, the structure to the base point. The inset orders the relaxation times of the three levels.
- **B.** The base manifold of structures with its fibers, the stationary section $\sigma_u$ and the resting section $\sigma_0$. Perception is vertical relaxation onto $\sigma_u$; learning is motion along the section.
- **C.** The probability simplex. Perturbing $p$ along an observable $f$ leaves the simplex; subtracting $\mathbb E_p[f]$ gives the centred tangent $f-\mathbb E_p[f]$, which lies in it.
- **D.** The convex log-normalizer $\psi(W)$: its slope is the correlation $C=\mathbb E[s\mu]$ (the Legendre-dual coordinate), its curvature the Fisher metric $g$.
